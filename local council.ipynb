```python
from pathlib import Path

# Create an output folder
output_folder = Path(r"C:\Users\madha\Downloads\Council_SQL_Output")
output_folder.mkdir(parents=True, exist_ok=True)

# Copy accepted data for export
sql_data = derived_accepted_data.copy()

# Convert Boolean columns from True/False to 1/0
bool_columns = sql_data.select_dtypes(include=['bool', 'boolean']).columns

sql_data[bool_columns] = sql_data[bool_columns].astype('Int64')

# Export accepted data for SQL
sql_data.to_csv(
    output_folder / "accepted_council_spend.csv",
    index=False,
    encoding="utf-8",
    date_format="%Y-%m-%d"
)

# Export rejected data
derived_rejected_data.to_csv(
    output_folder / "rejected_council_spend.csv",
    index=False,
    encoding="utf-8"
)

# Export review data
derived_review_data.to_csv(
    output_folder / "review_council_spend.csv",
    index=False,
    encoding="utf-8"
)

print("Files exported to:", output_folder)
```

```python
columns = [
    'transaction_id', 'transaction_date', 'council_name',
    'department', 'spend_category', 'supplier_name',
    'actual_spend_gbp', 'budget_amount_gbp',
    'spend_type', 'payment_status', 'procurement_method',
    'funding_source', 'currency', 'supplier_missing_name',
    'derived_supplier_name', 'derived_supplier_flag',
    'derived_payment_status', 'recorded_spend',
    'pending_spend', 'paid_spend', 'Variance_GBP',
    'Variance_pct', 'Supplier_Share_Of_Council_Pct'
]

if derived_accepted_data.columns.duplicated().any():
    raise ValueError("Resolve duplicate column names before exporting.")

data_to_load = derived_accepted_data[columns].copy()

flags = [
    'derived_supplier_flag', 'recorded_spend',
    'pending_spend', 'paid_spend'
]

data_to_load[flags] = data_to_load[flags].astype(int)

data_to_load.to_csv(
    'accepted_data.csv',
    index=False,
    date_format='%Y-%m-%d',
    lineterminator='\n',
    encoding='utf-8'
)

print("Expected rows:", len(data_to_load))
```

```python
csv_data = data_to_load.copy()

decimal_places = {
    'actual_spend_gbp': 2,
    'budget_amount_gbp': 2,
    'Variance_GBP': 2,
    'Variance_pct': 6,
    'Supplier_Share_Of_Council_Pct': 6
}

for column, places in decimal_places.items():
    csv_data[column] = csv_data[column].map(
        lambda value: f'{value:.{places}f}' if pd.notna(value) else ''
    )

csv_data.to_csv(
    'accepted_data.csv',
    index=False,
    date_format='%Y-%m-%d',
    lineterminator='\n',
    encoding='utf-8'
)
```
